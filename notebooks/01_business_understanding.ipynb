## 1. Business Problem

The ecommerce company has data about its products and order items. The data includes product categories, recorded item prices, freight values, product weights, and product dimensions.

The company needs to understand how its products and categories perform and how freight charges vary across products. Our task is to examine the available data, identify useful patterns, and provide recommendations supported by evidence.

**Main goal:** Understand product and category sales performance and investigate freight costs to help the business make better decisions.

We will only make conclusions that the available data can support.

- `product_length_cm`

### Order-items dataset

The `order-items` dataset contains:

- `product_id`
- `seller_id`
- `shipping_limit_date`
- `price`
- `freight_value`
- `product_category_name`
- `product_weight`

This dataset records order-item information, including product and seller identifiers, item price, freight value, and a shipping-limit date.

**Important:** We still need to confirm the exact meanings, units, and level of detail of some columns before calculating final metrics.

## 3. Relationship Between the Datasets

Both datasets contain `product_id`. We expect to use this column to join order-item records to product information.

Before joining, we must check:

- Whether `product_id` is unique in the `products` dataset.
- Whether every `product_id` in `order-items` has a matching product record.
- Whether either dataset contains duplicate records.
- Whether the join changes the number of order-item rows unexpectedly.

A one-to-many relationship is expected: one product may appear in many order-item records. We must verify this against the actual data before relying on it.

## 4. Business Objectives

1. Analyze recorded sales value by product and category.
2. Identify products and categories that contribute most to recorded sales.
3. Examine freight charges by product and category.
4. Investigate whether product weight or dimensions are associated with freight charges.
5. Explore seller-related patterns where the available data supports them.

## 5. Analytical Questions

### Product and category performance

1. Which product categories have the highest recorded sales value?
2. Which products have the highest recorded sales value?
3. What percentage of total recorded sales value comes from each category?
4. How many products are listed in each category?
5. How do recorded sales values vary by seller?

### Freight-cost analysis

1. Which products and categories have the highest total freight values?
2. What is the average freight value per order-item record?
3. What percentage of recorded sales value is represented by freight?
4. Is product weight associated with freight value?
5. Do product dimensions show a relationship with freight value?
6. Are the two weight columns consistent with each other?

### Data and reporting questions

1. How many distinct orders and order-item records are present?
2. Are there missing values, duplicate records, invalid values, or unmatched product IDs?
3. Can the available date fields support the time-based analysis required by the assignment?

## 6. Initial Business Metrics

We may calculate the following metrics after validating the data:

- **Recorded sales value:** Sum of `price`, if validation confirms that `price` is the item-level sales amount.
- **Distinct orders:** Count of unique `order_id` values.
- **Order-item records:** Number of records in the order-items dataset, after checking duplicates and the meaning of `order_item_id`.
- **Total freight value:** Sum of `freight_value`, after confirming its meaning and granularity.
- **Freight-to-sales ratio:** Total freight value divided by recorded sales value, multiplied by 100, if both measures are comparable.
- **Category sales contribution:** Category recorded sales value divided by total recorded sales value, multiplied by 100.

We must confirm the meaning and granularity of `price` and `freight_value` before finalizing these formulas.

**Not yet confirmed:** The listed columns do not include an explicit quantity column or an order purchase date. We should not assume we can calculate units sold or monthly sales trends from the listed columns alone. `shipping_limit_date` is not necessarily the purchase date.

**Profit limitation:** Revenue or recorded sales value is not the same as profit. The listed columns do not provide enough confirmed cost information to calculate actual profit.
