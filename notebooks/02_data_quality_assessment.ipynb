## Step 4: Check missing values

```python
# Check missing values in Products dataset
products_missing = pd.DataFrame({
    "Missing Count": products.isnull().sum(),
    "Missing Percentage": (products.isnull().sum()/len(products) * 100).round(2)
})

print(products_missing[products_missing["Missing Count"] > 0])
```

```python
# Check missing values in Order Items dataset
order_items_missing = pd.DataFrame({
    "Missing Count": order_items.isnull().sum(),
    "Missing Percentage": (order_items.isnull().sum()/len(order_items) * 100).round(2)
})

print(order_items_missing[order_items_missing["Missing Count"] > 0])
```
